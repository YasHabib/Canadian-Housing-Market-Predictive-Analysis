```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
```

### Pulling population data via web scraping from WorldPopulationReview.com and exploing that data

```python
url = "https://worldpopulationreview.com/canadian-provinces"
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

Upon inspecting the webpage, we find that the table we are looking to scrape is under the Div "m-0 rounded-xl border border-wpr-table_border max-h-[60vh] overflow-auto p-0" and the table class is "wpr-table !border-none"

```python
table_div = soup.find('div', class_='m-0 rounded-xl border border-wpr-table_border max-h-[60vh] overflow-auto p-0')
table = table_div.find('table', class_='wpr-table !border-none')
```

```python
if table is None:
    print("Table not found")
else:
    # Extract the header of the table
    header_row = table.find_all('th')
    headers = [header.text.strip() for header in header_row[:6]]  # Only keep the first 6 columns as headers
    print("Headers:")
    print(headers)
    
    
    # Extract all the rows of the table
    rows = []
    for row in table.find_all('tr')[1:]:  # Skip the header row
        cols = row.find_all('td')

        # Check if the row has data
        if cols:
            # Extract the province name from the first column's <a> tag
            province = row.find('th').find('a').text.strip()  # Province is in the first <th> column with <a> tag
            data = [province] + [col.text.strip() for col in cols]  # Add province name to the data

            # Skip rows that don't match the number of columns
            if len(data) == len(headers):
                rows.append(data)

    # Print out the rows
    print("\nRows:")
    for r in rows:
        print(r)

    # Create a DataFrame for easy data manipulation
    population_df = pd.DataFrame(rows, columns=headers)

    # Print the DataFrame to check the scraped data
    print(population_df)
```

```python
# Convert numeric columns (population and change values) to integers and remove commas
for col in population_df.columns[1:]:  # Skip the 'Province' column
    population_df[col] = population_df[col].replace({',': '', '%':''}, regex=True)  # Remove commas
    population_df[col] = pd.to_numeric(population_df[col], errors='coerce')  # Convert to numeric, coercing errors to NaN

# Print the cleaned DataFrame
print(population_df)
```

### Populating Cost of living index. (Consumer Price Index, for reference, the index at 2002 was 100)
##### Source:  https://www150.statcan.gc.ca/t1/tbl1/en/cv.action?pid=1810000501

```python
cpi_df = pd.read_csv('Pulled data//CostOfLivingByProvinceAndCategory.csv', header=[0,1], index_col=0)
```

```python
cpi_df.head()
```

```python
#Looking at the CSV, we have noticed that there are multiple Unnamed:2_level_0 like headers besides Shelter. We will need to clean it out.
#Easiest way of doing it is to remove the top header row and flattening the multi-level column index.

cpi_df.columns = cpi_df.columns.droplevel(0)
```

```python
# Moving the "Geography" to column as it is currently in index.
cpi_df.reset_index(inplace=True)
cpi_df.rename(columns={"index": "Geography"}, inplace=True)  # Rename the default index column
print(cpi_df.columns.tolist())

# First, we will remove Nanuvat as we do not have any data for it and we will not be predicting it's housing cost.
cpi_df = cpi_df[cpi_df["Geography"] != "Iqaluit, Nunavut 7"]

# Reset the index after removal for a clean DataFrame
cpi_df = cpi_df.reset_index(drop=True) 

# Display the updated DataFrame
cpi_df
```

##### Given that we are missing 2024 data, we will try to populate it using regression. First, we will plot the data to see it's trend.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming your cpi_df is already loaded as a DataFrame
# If you haven't loaded it, you can use the following:
# cpi_df = pd.read_csv('path_to_your_data.csv')

# Reshaping the DataFrame for easier plotting
cpi_df = cpi_df.melt(id_vars=["Geography"], var_name="Year", value_name="CPI")

# Convert Year to an integer type
cpi_df['Year'] = cpi_df['Year'].astype(int)

# Plotting the CPI trends for each province/territory
plt.figure(figsize=(14, 8))
sns.lineplot(data=cpi_df, x='Year', y='CPI', hue='Geography', marker='o')

# Customize the plot
plt.title('CPI Trends Over Time by Province/Territory', fontsize=16)
plt.xlabel('Year', fontsize=14)
plt.ylabel('CPI', fontsize=14)
plt.xticks(rotation=45)
plt.legend(title='Geography', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Importing ML library
from sklearn.linear_model import LinearRegression
```

```python
# Prepare years for regression
years = np.array(range(2011, 2024)).reshape(-1, 1)  # 2011 to 2023

cpi_df["2024"] = np.nan

# Loop through each province
for index, row in cpi_df.iterrows():
    cpi_values = row[1:-1].values  # Select CPI values (excluding Geography and 2024)
    model = LinearRegression()
    model.fit(years, cpi_values)  # Fit model
    cpi_2024 = model.predict([[2024]])  # Predict for 2024
    cpi_df.at[index, "2024"] = cpi_2024[0]  # Add prediction to the DataFrame

# Display the DataFrame with 2024 predictions
print(cpi_df)
```
